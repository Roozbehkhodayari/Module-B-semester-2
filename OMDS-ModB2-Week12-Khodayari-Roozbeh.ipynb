# 2. Weekly graph question

Suppose you plot the number visitors to a museum over the seven days of a week; then again the following week.  Can you suggest the advantages and disadvantages of the following two ways of presenting the data?

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
time_series_0 = np.array([100, 110, 130, 120, 100, 90, 80])
time_series_1 = np.array([70, 90, 100, 100, 120, 110, 90])
fig, axs = plt.subplots(1, 2)
axs[0].plot(time_series_0)
axs[1].plot(time_series_1)
plt.subplots_adjust(wspace = 0)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

```python
plt.plot(time_series_0)
plt.plot(time_series_1)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

The first method shows the two weeks in separate plots. This makes each week easy to see by itself. However, it is harder to compare the two weeks directly because the viewer has to look back and forth between the two plots. Also, if the y-axis scale is not the same in both plots, the comparison can be misleading.

The second method puts both weeks on the same plot. This makes comparison easier because both lines share the same x-axis and y-axis. It is easier to see which week had more visitors on each day and where the patterns are different.

The disadvantage of the second method is that it can become messy if there are many lines. It also needs a clear legend or labels so the viewer knows which line represents each week.

Overall, I think the second method is better for comparison because the main goal is to compare the two weeks. The first method is better if we want to study each week separately.

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 187-205 of the Storytelling With Data book as best you can. (Chapter Eight). You do not have to get the exact data values right, just the overall look and feel.

```python
product = ["A"] * 7 + ["B"] * 7 + ["C"] * 5 + ["D"] * 4 + ["E"] * 2
year = np.concatenate([np.arange(2008, 2015), np.arange(2008, 2015),np.arange(2010, 2015),np.arange(2011, 2015),np.arange(2013, 2015)])
price = [395,415,420,390, 300, 270,260,360,400, 410,370, 290, 260,250, 100, 185, 200, 250, 185, 170, 270, 215, 210, 100, 210]
data = pd.DataFrame({"product": product, "year": year, "price": price})

line_color = "gray"
range_color = "lightblue"
text_blue = "navy"

fig, ax = plt.subplots(figsize=(8, 4.8))

# Recommended range
ax.fill_between([2007, 2014], 150, 200, color=range_color, alpha=0.45)
ax.text(2007.15, 170, "Recommended range", color=text_blue, fontsize=12, va="center")

# Plot one line for each product
for p, group in data.groupby("product"):
    ax.plot(group["year"], group["price"], color=line_color, linewidth=2.2)

# Left-side labels for A and B
ax.text(2007.75, 395, "A", color=line_color, fontsize=11, ha="left", va="center")
ax.text(2007.75, 355, "B", color=line_color, fontsize=11, ha="left", va="center")

# Middle / lower labels for C, D, E
ax.text(2009.85, 100, "C", color=line_color, fontsize=11, ha="center", va="center")
ax.text(2010.90, 140, "D", color=line_color, fontsize=11, ha="center", va="center")
ax.text(2012.90, 95,  "E", color=line_color, fontsize=11, ha="center", va="center")

# Average point at 2014
last_year = data[data["year"] == 2014]
avg_2014 = last_year["price"].mean()

ax.scatter(2014, avg_2014, color=text_blue, s=22, zorder=5)
ax.text(2013.92, avg_2014 + 6, "AVG", color=text_blue, fontsize=10, ha="right", va="center")

ax.set_title("Retail price over time", loc="left", fontsize=16, y=1.07, x=-0.1)
ax.set_xlabel("Year", fontsize=11)
ax.xaxis.set_label_coords(0.14, -0.1)
ax.set_ylabel("Average price", fontsize=11, loc="top")

ax.set_xlim(2007, 2014.2)
ax.set_ylim(0, 500)

ax.set_xticks(np.arange(2008, 2015))
ax.tick_params(axis="x", labelsize=10)

y_ticks = [0, 150, 200, 300, 400, 500]
ax.set_yticks(y_ticks)
ax.set_yticklabels([f"${y}" for y in y_ticks])

# Clean up spines
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["left"].set_color("lightgray")
ax.spines["bottom"].set_color("lightgray")

ax.tick_params(colors="dimgray")

plt.tight_layout()
plt.show()
```
